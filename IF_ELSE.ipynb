```python
# DECISION MAKING IN PYTHON 

IF - ELSE  statement provides an else block combined with the if statement 

if the condition is True, then the if block is executed , otherwise the else block is executed. 

ELIF statement works like an if-else-if ladder statements 
```

```python
age =int(input('enter your age'))
if age >=18:
    print("Eligible to vote:",age)

else:
    print("Not eligible to vote")
```

```python
# traffic signal
# red = stop
# yellow = wait
# green = go

color =input("enter the color")
          
if color == "red":
    print ('Stop')
elif color == "yellow":
    print ('wait')
elif color == "green":
    print('go')
else :
    print('invalid color')
```

```python

# NESTED IF

# role login

username = input('Enter your username:').lower()

if username == 'admin':
    role = input('enter your role').lower()
    if role == 'manager':
        print('Full access')
    elif username == 'employee':
        print('limited access')
    else:
        print('unknown role')

else:
    print('unknown user')
```
