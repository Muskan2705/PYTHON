```python
ce=0
for i in range(1,11):
    if i%2==0:
        ce = ce+1

print("count of even", ce)
```

```python
even_number = 0
odd_number = 0
even_sum = 0
odd_sum = 0

for i in range(1,11):
    if i%2 == 0:
        even_number = even_number+1
        even_sum += i

    else:
        odd_number = odd_number+1
        odd_sum =+ i
print('count of even',even_number)
print('count of odd',odd_number)
print('sum of even',even_sum)
print('sum of even',odd_sum)
```

```python
# do not use reverse 
#  palindrome 
t = input('enter your word')
e_s = ''

for i in t:
    e_s = i +e_s

if t ==e_s:
    print('yes,it is palindrome')

else:
    print('no,it is not')

```

```python
# vowels count 
# consonents count 
# spaces count
a = input('enter your word').lower()

vowels = 0
consonants = 0
spaces = 0

for i in a:
    if i in 'aeiou':
        vowels+=1
    elif i==' ':
        spaces+=1
    else:
        consonants+=1
print(vowels,consonants,spaces)        
```

```python
# count of digits
# count of alphabets 
# count of special charcater 
# space count nhi hona chaiye

a = input('Enter your word')

digits = 0
alphabets = 0
special_character = 0

for i in a:
    if i =='':
        continue
    elif i.isdigit():
        digits +=1
    elif i.isalpha():
        alphabets +=1
    else:
        special_character +=1
print(digits)
print(alphabets)
print(special_character)
```
