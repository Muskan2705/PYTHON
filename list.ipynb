```python
# list comprehension 
```

```python
p =[]
for i in range(10):
    if i%2==0:
        p.append(i)
p        
```

```python
k = [a for a in range(10) if a %2==0]
k
```

```python
a = [x for x in range(5) if x!=2]
a
```

```python
[word for word in ["apple","ant","ball"]if word.startswith('a')]
```

```python
b = [x if x%2==0 else "odd" for x in range(5)]
b
```

```python
c = [len(word) for word in ["cat","ball","python","AI"]]
c
```

```python
d = [x*y for x in range (1,4) for y in range(1,4)]
d
```
