```python
# anagram code
a = input('enter your word').lower()
b = input('enter your word').lower()

is_anagram = True

if len(a) == len(b):
    # print('not anagram')
    for i in a:
        if a.count(i) != b.count(i):
            is_anagram = False 


else:
    is_anagram = False

if is_anagram :
    print('yes, it is a anagram')

else:
    print('no, it is not anagram')
```
